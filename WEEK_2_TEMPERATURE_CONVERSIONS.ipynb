#TEMPERATURE CONVERSION PROGRAM#
this programme converts temperature from celsius to fahrenheight,
checks whether temperature is above or below a specified threshold,
and uses a loop to convert several temeperatures.

```python
def celsius_to_fahrenheit(celsius) :
    fahrenheit = (celsius * 9/5) + 32
    return fahrenheit
  #function to convert celcius to fahrenheit#
```

```python
celsius = float(input("enter temperature in celsius: "))
fahrenheit = celsius_to_fahrenheit(celsius)
print("Temperature in fahrenheit:" , fahrenheit)
#print(Temperature in Fahrenheit)
```

```python
if celsius >= 25:
  print("Above the threshold")
else:
  print("Below the threshold")
#Check the threshold#
```

```python
for celsius in[10, 20, 30, 40]:
    print(celsius, "celsius=", celsius_to_fahrenheit(celsius), "fahrenheit")
#Convert several temperatures#
```
